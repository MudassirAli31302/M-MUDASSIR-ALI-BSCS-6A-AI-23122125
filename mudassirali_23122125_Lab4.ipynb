```python

!pip install -U google-genai

from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

MODEL = "models/gemini-3.5-flash"

department_info = {
    "department": "Computer Science (FUUAST)",
    "timings": "9:00 AM to 3:40 PM (Monday to Friday)",
    "lecture_duration": "50 minutes per class",
    "break_time": "1:10 PM to 2:00 PM",
    "rooms_and_labs": "Lab 1 to 4, Room 1, 2, 3, 11, 12, 13, Hall",
    "semesters_courses": {
        "BS 2nd Semester": "OOP, Digital Logic Design (DLD), Pre-Calculus, Pak Studies, Arts and Humanities, Quantitative Reasoning, Fehm-e-Quran",
        "BS 4th Semester": "Database (DBMS), Theory of Automata, Linear Algebra, Design and Analysis of Algorithms, Accounting",
        "BS 6th Semester": "Artificial Intelligence (AI), Natural Language Processing (NLP), Compiler Construction, Numerical Computing, Professional Practices",
        "BS 7th Semester": "Data Science, Parallel and Distributed Computing, Cloud Computing, Entrepreneurship",
        "BS 8th Semester": "Human Computer Interaction (HCI), Wireless Networks"
    }
}

def ask_agent(query):
    prompt = f"""
You are a helpful student support assistant for the FUUAST Computer Science Department.

Department Information:
- Department: {department_info['department']}
- Working Hours: {department_info['timings']}
- Class Duration: {department_info['lecture_duration']}
- Break: {department_info['break_time']}
- Available Rooms and Labs: {department_info['rooms_and_labs']}
- Semester Courses: {department_info['semesters_courses']}

Guidelines:
1. Answer only questions about the CS department, courses, and timetable.
2. Use a friendly and informal tone.
3. Give short and clear answers in simple language.

Student Query: "{query}"

Your Answer:
"""

    result = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return result.text


print("=== FUUAST CS Student Help Desk ===")
print("Type 'exit' to close the assistant.\n")

while True:
    message = input("You: ")

    if message.strip().lower() == "exit":
        print("Thank you for using the CS Help Desk. Goodbye!")
        break

    try:
        answer = ask_agent(message)
        print("\nHelp Desk:\n", answer, "\n")

    except Exception as error:
        print("Error occurred:", error)
```
